## Task 1: Create a Connection

Create an in-memory DuckDB connection.

**Hint:** Use `duckdb.connect()`

```python
# Solution
conn = duckdb.connect(':memory:')
print(f"Connected to DuckDB {duckdb.__version__}")
```

## Task 2: Create Your First Table

Create a table called `students` with the following columns:
- `id` (INTEGER)
- `name` (VARCHAR)
- `age` (INTEGER)
- `major` (VARCHAR)

Add at least 5 students.

```python
# Solution
conn.execute("""
    CREATE TABLE students (
        id INTEGER PRIMARY KEY,
        name VARCHAR,
        age INTEGER,
        major VARCHAR
    )
""")

conn.execute("""
    INSERT INTO students VALUES
    (1, 'Alice', 21, 'Computer Science'),
    (2, 'Bob', 23, 'Mathematics'),
    (3, 'Carol', 20, 'Computer Science'),
    (4, 'David', 25, 'Physics'),
    (5, 'Eva', 22, 'Mathematics'),
    (6, 'Frank', 24, 'Computer Science'),
    (7, 'Grace', 21, 'Physics')
""")

print("Table 'students' created with 7 entries")
```

## Task 3: Query Data

a) Show all students

```python
# Solution
conn.execute("SELECT * FROM students").df()
```

b) Show only names and majors of all students

```python
# Solution
conn.execute("SELECT name, major FROM students").df()
```

c) Find all students older than 22

```python
# Solution
conn.execute("SELECT * FROM students WHERE age > 22").df()
```

## Task 4: Aggregations

a) How many students are there in total?

```python
# Solution
conn.execute("SELECT COUNT(*) AS total_students FROM students").df()
```

b) What is the average age of all students?

```python
# Solution
conn.execute("SELECT ROUND(AVG(age), 1) AS average_age FROM students").df()
```

c) How many students are there per major?

```python
# Solution
conn.execute("""
    SELECT major, COUNT(*) AS student_count 
    FROM students 
    GROUP BY major
    ORDER BY student_count DESC
""").df()
```

## Bonus Task

Find the youngest and oldest student in each major.

```python
# Solution
conn.execute("""
    SELECT 
        major,
        MIN(age) AS youngest_age,
        MAX(age) AS oldest_age,
        MAX(age) - MIN(age) AS age_range
    FROM students
    GROUP BY major
    ORDER BY major
""").df()
```
