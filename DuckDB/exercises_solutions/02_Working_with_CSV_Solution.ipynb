## Task 1: Query CSV Directly

Read the file `../sample_data/states.csv` directly without creating a table.

**Hint:** Use `read_csv_auto()`

```python
# Solution
conn.execute("SELECT * FROM read_csv_auto('../sample_data/states.csv')").df()
```

## Task 2: Create Table from CSV

Create a table `states` from the CSV file.

```python
# Solution
conn.execute("""
    CREATE TABLE states AS 
    SELECT * FROM read_csv_auto('../sample_data/states.csv')
""")
print("Table 'states' created")
```

## Task 3: Data Exploration

a) How many rows does the table have?

```python
# Solution
conn.execute("SELECT COUNT(*) AS row_count FROM states").df()
```

b) Show the first 10 rows

```python
# Solution
conn.execute("SELECT * FROM states LIMIT 10").df()
```

c) What columns exist? Use `DESCRIBE`

```python
# Solution
conn.execute("DESCRIBE states").df()
```

## Task 4: AdventureWorks Data

Load the following CSV files from `../sample_data/AW_CSV/`:
- `Sales.Customer.csv` as table `customers`
- `Sales.SalesTerritory.csv` as table `territories`

```python
# Solution
conn.execute("""
    CREATE TABLE customers AS 
    SELECT * FROM read_csv_auto('../sample_data/AW_CSV/Sales.Customer.csv')
""")

conn.execute("""
    CREATE TABLE territories AS 
    SELECT * FROM read_csv_auto('../sample_data/AW_CSV/Sales.SalesTerritory.csv')
""")

print("Tables 'customers' and 'territories' created")
```

## Task 5: Analyze Data

a) How many customers are there in total?

```python
# Solution
conn.execute("SELECT COUNT(*) AS total_customers FROM customers").df()
```

b) How many different territories exist?

```python
# Solution
conn.execute("SELECT COUNT(*) AS territory_count FROM territories").df()
```

c) List all territory names, sorted alphabetically

```python
# Solution
conn.execute("""
    SELECT Name AS territory_name 
    FROM territories 
    ORDER BY Name
""").df()
```

## Task 6: Export Data

Export the territory list as a CSV file to `../exports/my_territories.csv`

**Hint:** Use `COPY ... TO`

```python
# Solution
conn.execute("""
    COPY (
        SELECT Name AS territory_name 
        FROM territories 
        ORDER BY Name
    ) TO '../exports/my_territories.csv' (HEADER, DELIMITER ',')
""")

print("Exported to ../exports/my_territories.csv")
```

## Bonus Task

Load ALL CSV files from the folder `../sample_data/AW_CSV/` with a single command.

**Hint:** Use wildcards `*.csv`

```python
# Solution
result = conn.execute("""
    SELECT filename, COUNT(*) AS row_count
    FROM read_csv_auto('../sample_data/AW_CSV/*.csv', filename=true)
    GROUP BY filename
    ORDER BY filename
""").df()

print(f"Loaded {len(result)} CSV files:")
result
```
